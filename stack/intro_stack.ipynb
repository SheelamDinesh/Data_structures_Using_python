```python
class My_stack:
    def __init__(self, capacity):
        self.top = -1
        self.capacity = capacity
        self.arr = [0] * capacity

    def push(self, data):
        if self.top == self.capacity - 1:
            return "stack is Full"
        self.top+=1
        self.arr[self.top] = data


    def pop(self):
        if self.top == -1:
            return "Stack is Eampty"
        return self.arr[-1]
        
    def size(self):
        return len(self.arr)
```

```python
stack1 = My_stack(2)
```

```python
stack1.push(1)
```

```python
print(stack1.arr)
```

```python
stack1.push(3)
```

```python
print(stack1.arr)
```

```python
stack1.size()
```
