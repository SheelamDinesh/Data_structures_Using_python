```python
class Node:
    def __init__(self, data):
        self.left = None
        self.right = None
        self.data = data
```

```python
root_node = Node(1)
n2 = Node(2)
n3 = Node(3)
n4 = Node(4)
n5 = Node(5)


root_node.left = n2
root_node.right = n3
n2.left = n4
n2.right = n5


def binary_tree_traversal(root_node):
    if root_node is None or root_node.left is None or root_node.right is None:
        return 

    left_node = root_node.left
    print(left_node.data)
    right_node = root_node.right
    print(right_node.data)
    binary_tree_traversal(left_node)
    binary_tree_traversal(right_node)



binary_tree_traversal(root_node)
```

```python
binary_tree_traversal(root_node)
```
