Breadth First search

```python
class Node:
    def __init__(self, data):
        self.left = None
        self.right = None
        self.data = data

root_node = Node(1)
n2 = Node(2)
n3 = Node(3)
n4 = Node(4)
n5 = Node(5)


root_node.left = n2
root_node.right = n3
n2.left = n4
n2.right = n5
```

```python
from collections import deque

def binary_tree_traversal(root):
    if root is None:
        return
    queue = deque()
    queue.append(root)
    while queue:
        root = queue.popleft()
        print(root.data,end = " ")
        if root.left is not None:
            queue.append(root.left)
        if root.right is not None:
            queue.append(root.right)
  

binary_tree_traversal(root_node)
```
