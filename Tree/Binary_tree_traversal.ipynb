```python
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import mm
from reportlab.lib import colors
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
    HRFlowable, KeepTogether, PageBreak
)
from reportlab.lib.enums import TA_LEFT, TA_CENTER
from reportlab.platypus import Flowable

PAGE_W, PAGE_H = A4
MARGIN = 15 * mm

# ── Color palette ──────────────────────────────────────────────────────────
C_BG_HEADER   = colors.HexColor("#1a1a2e")   # deep navy – main title bg
C_BG_SECTION  = colors.HexColor("#16213e")   # section header bg
C_BG_CAT      = colors.HexColor("#0f3460")   # category header bg
C_ACCENT      = colors.HexColor("#e94560")   # accent / star badges
C_TEXT_LIGHT  = colors.HexColor("#ffffff")
C_TEXT_DARK   = colors.HexColor("#1a1a2e")
C_CHECKBOX_BG = colors.HexColor("#f5f7ff")
C_BORDER      = colors.HexColor("#c5cae9")
C_BEGINNER    = colors.HexColor("#43a047")
C_INTER       = colors.HexColor("#fb8c00")
C_ADVANCED    = colors.HexColor("#e53935")
C_ROW_ALT     = colors.HexColor("#eef0fb")
C_NOTE_BG     = colors.HexColor("#fff8e1")
C_NOTE_BORDER = colors.HexColor("#ffd54f")

# ── Small checkbox flowable ─────────────────────────────────────────────────
class Checkbox(Flowable):
    def __init__(self, size=4.5*mm):
        super().__init__()
        self.size = size
        self.width = size
        self.height = size
    def draw(self):
        s = self.size
        self.canv.setFillColor(colors.white)
        self.canv.setStrokeColor(C_BORDER)
        self.canv.setLineWidth(0.8)
        self.canv.roundRect(0, 0, s, s, 1*mm, fill=1, stroke=1)

# ── Helpers ─────────────────────────────────────────────────────────────────
def badge(text, bg):
    return f'<font color="white"><b>{text}</b></font>'

def level_cell(level):
    if level == "B":
        return colors.HexColor("#e8f5e9"), "Beginner", C_BEGINNER
    if level == "I":
        return colors.HexColor("#fff3e0"), "Intermediate", C_INTER
    return colors.HexColor("#ffebee"), "Advanced", C_ADVANCED

def make_styles():
    base = getSampleStyleSheet()
    s = {}

    s['doc_title'] = ParagraphStyle("doc_title",
        fontName="Helvetica-Bold", fontSize=22, textColor=C_TEXT_LIGHT,
        alignment=TA_CENTER, leading=28, spaceAfter=4)

    s['doc_sub'] = ParagraphStyle("doc_sub",
        fontName="Helvetica", fontSize=10, textColor=colors.HexColor("#b0bec5"),
        alignment=TA_CENTER, leading=14)

    s['section_hdr'] = ParagraphStyle("section_hdr",
        fontName="Helvetica-Bold", fontSize=13, textColor=C_TEXT_LIGHT,
        leading=18, leftIndent=4)

    s['cat_hdr'] = ParagraphStyle("cat_hdr",
        fontName="Helvetica-Bold", fontSize=10.5, textColor=C_TEXT_LIGHT,
        leading=15, leftIndent=4)

    s['topic'] = ParagraphStyle("topic",
        fontName="Helvetica", fontSize=9, textColor=C_TEXT_DARK,
        leading=13, leftIndent=2)

    s['topic_bold'] = ParagraphStyle("topic_bold",
        fontName="Helvetica-Bold", fontSize=9, textColor=C_TEXT_DARK,
        leading=13, leftIndent=2)

    s['note'] = ParagraphStyle("note",
        fontName="Helvetica-Oblique", fontSize=8.5, textColor=colors.HexColor("#5d4037"),
        leading=12, leftIndent=4, rightIndent=4)

    s['label_b'] = ParagraphStyle("label_b",
        fontName="Helvetica-Bold", fontSize=7.5, textColor=C_BEGINNER,
        alignment=TA_CENTER, leading=10)
    s['label_i'] = ParagraphStyle("label_i",
        fontName="Helvetica-Bold", fontSize=7.5, textColor=C_INTER,
        alignment=TA_CENTER, leading=10)
    s['label_a'] = ParagraphStyle("label_a",
        fontName="Helvetica-Bold", fontSize=7.5, textColor=C_ADVANCED,
        alignment=TA_CENTER, leading=10)

    s['footer'] = ParagraphStyle("footer",
        fontName="Helvetica-Oblique", fontSize=7.5, textColor=colors.grey,
        alignment=TA_CENTER)

    s['role_title'] = ParagraphStyle("role_title",
        fontName="Helvetica-Bold", fontSize=10, textColor=C_BG_SECTION,
        leading=14)

    s['role_body'] = ParagraphStyle("role_body",
        fontName="Helvetica", fontSize=8.5, textColor=C_TEXT_DARK,
        leading=13, leftIndent=6)

    s['week_hdr'] = ParagraphStyle("week_hdr",
        fontName="Helvetica-Bold", fontSize=11, textColor=C_TEXT_LIGHT,
        leading=16, leftIndent=4)

    return s

# ── Build section header ─────────────────────────────────────────────────────
def section_header(title, styles, emoji=""):
    p = Paragraph(f"{emoji}  {title}" if emoji else title, styles['section_hdr'])
    tbl = Table([[p]], colWidths=[PAGE_W - 2*MARGIN])
    tbl.setStyle(TableStyle([
        ("BACKGROUND", (0,0), (-1,-1), C_BG_SECTION),
        ("TOPPADDING",    (0,0), (-1,-1), 6),
        ("BOTTOMPADDING", (0,0), (-1,-1), 6),
        ("LEFTPADDING",   (0,0), (-1,-1), 10),
        ("ROUNDEDCORNERS", [3]),
    ]))
    return tbl

def cat_header(title, styles):
    p = Paragraph(title, styles['cat_hdr'])
    tbl = Table([[p]], colWidths=[PAGE_W - 2*MARGIN])
    tbl.setStyle(TableStyle([
        ("BACKGROUND", (0,0), (-1,-1), C_BG_CAT),
        ("TOPPADDING",    (0,0), (-1,-1), 4),
        ("BOTTOMPADDING", (0,0), (-1,-1), 4),
        ("LEFTPADDING",   (0,0), (-1,-1), 10),
    ]))
    return tbl

# ── Checklist rows: (topic_text, level "B"/"I"/"A", notes="") ───────────────
def checklist_table(items, styles):
    """items = list of (text, level, note_optional)"""
    COL_CB   = 8*mm
    COL_LVL  = 22*mm
    COL_TEXT = PAGE_W - 2*MARGIN - COL_CB - COL_LVL

    rows = []
    for idx, item in enumerate(items):
        text  = item[0]
        level = item[1]          # B / I / A
        note  = item[2] if len(item) > 2 else ""

        bg_row, lbl_text, lbl_color = level_cell(level)

        lbl_style = styles[f'label_{"b" if level=="B" else "i" if level=="I" else "a"}']
        topic_p = Paragraph(text, styles['topic'])
        lbl_p   = Paragraph(lbl_text, lbl_style)

        cell_content = [topic_p]
        if note:
            cell_content.append(Paragraph(f"↳ {note}", styles['note']))

        row = [Checkbox(), cell_content, lbl_p]
        rows.append(row)

    tbl = Table(rows, colWidths=[COL_CB, COL_TEXT, COL_LVL], repeatRows=0)

    style_cmds = [
        ("VALIGN",        (0,0), (-1,-1), "MIDDLE"),
        ("TOPPADDING",    (0,0), (-1,-1), 3.5),
        ("BOTTOMPADDING", (0,0), (-1,-1), 3.5),
        ("LEFTPADDING",   (0,0), (-1,-1), 4),
        ("RIGHTPADDING",  (0,0), (-1,-1), 4),
        ("GRID",          (0,0), (-1,-1), 0.4, C_BORDER),
        ("ROWBACKGROUNDS",(0,0), (-1,-1), [C_CHECKBOX_BG, C_ROW_ALT]),
    ]
    tbl.setStyle(TableStyle(style_cmds))
    return tbl

def note_box(text, styles):
    p = Paragraph(f"📌  {text}", styles['note'])
    tbl = Table([[p]], colWidths=[PAGE_W - 2*MARGIN])
    tbl.setStyle(TableStyle([
        ("BACKGROUND",    (0,0), (-1,-1), C_NOTE_BG),
        ("LEFTPADDING",   (0,0), (-1,-1), 8),
        ("RIGHTPADDING",  (0,0), (-1,-1), 8),
        ("TOPPADDING",    (0,0), (-1,-1), 5),
        ("BOTTOMPADDING", (0,0), (-1,-1), 5),
        ("BOX",           (0,0), (-1,-1), 1, C_NOTE_BORDER),
    ]))
    return tbl

# ═══════════════════════════════════════════════════════════════════════════
#  MAIN
# ═══════════════════════════════════════════════════════════════════════════
def build_pdf(path):
    doc = SimpleDocTemplate(
        path, pagesize=A4,
        leftMargin=MARGIN, rightMargin=MARGIN,
        topMargin=MARGIN, bottomMargin=MARGIN,
    )
    styles = make_styles()
    story  = []
    SP     = lambda n=4: Spacer(1, n*mm)

    # ── COVER HEADER ──────────────────────────────────────────────────────
    header_data = [[
        Paragraph("DINESH SHEELAM — AI/ML SKILL MASTERY CHECKLIST", styles['doc_title']),
    ]]
    header_tbl = Table(header_data, colWidths=[PAGE_W - 2*MARGIN])
    header_tbl.setStyle(TableStyle([
        ("BACKGROUND",    (0,0), (-1,-1), C_BG_HEADER),
        ("TOPPADDING",    (0,0), (-1,-1), 12),
        ("BOTTOMPADDING", (0,0), (-1,-1), 8),
        ("LEFTPADDING",   (0,0), (-1,-1), 10),
        ("ROUNDEDCORNERS", [4]),
    ]))
    story.append(header_tbl)

    sub_data = [[Paragraph(
        "2-Week Structured Preparation Plan  •  Beginner → Expert Coverage  •  2025-2026 Market Aligned",
        styles['doc_sub'])]]
    sub_tbl = Table(sub_data, colWidths=[PAGE_W - 2*MARGIN])
    sub_tbl.setStyle(TableStyle([
        ("BACKGROUND",    (0,0), (-1,-1), colors.HexColor("#12122a")),
        ("TOPPADDING",    (0,0), (-1,-1), 5),
        ("BOTTOMPADDING", (0,0), (-1,-1), 8),
        ("LEFTPADDING",   (0,0), (-1,-1), 10),
    ]))
    story.append(sub_tbl)
    story.append(SP(5))

    # ── LEGEND ────────────────────────────────────────────────────────────
    legend = Table([[
        Paragraph("🟢  Beginner", styles['label_b']),
        Paragraph("🟠  Intermediate", styles['label_i']),
        Paragraph("🔴  Advanced/Expert", styles['label_a']),
        Paragraph("☐  = Check when confident", styles['note']),
    ]], colWidths=[(PAGE_W - 2*MARGIN)/4]*4)
    legend.setStyle(TableStyle([
        ("BOX",           (0,0), (-1,-1), 0.5, C_BORDER),
        ("INNERGRID",     (0,0), (-1,-1), 0.3, C_BORDER),
        ("VALIGN",        (0,0), (-1,-1), "MIDDLE"),
        ("TOPPADDING",    (0,0), (-1,-1), 4),
        ("BOTTOMPADDING", (0,0), (-1,-1), 4),
    ]))
    story.append(legend)
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 1 — PYTHON
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("1.  PYTHON FOR AI/ML", styles, "🐍"))
    story.append(SP(2))

    story.append(cat_header("Core Language Fundamentals", styles))
    story.append(checklist_table([
        ("Variables, Data Types (int, float, str, bool, None)", "B"),
        ("Lists, Tuples, Sets, Dictionaries — CRUD & iteration", "B"),
        ("List/Dict/Set comprehensions", "B"),
        ("Functions — args, kwargs, default params, return values", "B"),
        ("Lambda functions & map/filter/zip", "B"),
        ("Modules & packages — import, __init__.py", "B"),
        ("Exception handling — try/except/finally, custom exceptions", "B"),
        ("File I/O — open/read/write/CSV/JSON", "B"),
        ("String methods & f-strings (format, split, strip, replace)", "B"),
        ("Conditional logic & loops (for, while, break, continue)", "B"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Object-Oriented Programming (OOP)", styles))
    story.append(checklist_table([
        ("Classes & Objects — __init__, self, attributes", "B"),
        ("Instance methods vs class methods vs static methods", "I"),
        ("Inheritance & method overriding", "I"),
        ("Encapsulation — public, protected, private attributes", "I"),
        ("Polymorphism & duck typing", "I"),
        ("Magic/Dunder methods — __str__, __repr__, __len__, __eq__", "I"),
        ("Abstract classes (ABC) & interfaces", "I"),
        ("Dataclasses (@dataclass decorator)", "I"),
        ("Composition vs Inheritance patterns", "A"),
        ("Design patterns relevant to ML — Factory, Strategy, Observer", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Intermediate & Advanced Python", styles))
    story.append(checklist_table([
        ("Decorators — @staticmethod, @classmethod, custom decorators", "I"),
        ("Generators & iterators — yield, next(), iter()", "I"),
        ("Context managers — with statement, __enter__/__exit__", "I"),
        ("*args and **kwargs unpacking deeply", "I"),
        ("Type hints — int, str, List, Dict, Optional, Union (typing module)", "I"),
        ("Virtual environments — venv, pip, requirements.txt", "I"),
        ("Async / Await — asyncio basics for API calls", "I"),
        ("Logging module — levels, handlers, formatters", "I"),
        ("Unit testing — unittest, pytest basics", "I"),
        ("Profiling & memory optimization — cProfile, memory_profiler", "A"),
        ("Multiprocessing vs multithreading — GIL awareness", "A"),
        ("Python packaging — setup.py, pyproject.toml", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 2 — DATA & MATH
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("2.  DATA MANIPULATION & MATH", styles, "📊"))
    story.append(SP(2))

    story.append(cat_header("NumPy", styles))
    story.append(checklist_table([
        ("Array creation — np.array, zeros, ones, arange, linspace", "B"),
        ("Array shapes — reshape, flatten, squeeze, expand_dims", "B"),
        ("Indexing & slicing — basic, fancy, boolean masking", "B"),
        ("Broadcasting rules", "I"),
        ("Universal functions (ufuncs) — np.sum, mean, std, max, min", "B"),
        ("Matrix operations — dot product, matmul, transpose", "I"),
        ("Linear algebra — np.linalg (eig, svd, inv, norm)", "I"),
        ("Random module — np.random.seed, randn, randint, choice", "B"),
        ("Stacking — vstack, hstack, concatenate", "I"),
        ("Performance: vectorization vs Python loops", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Pandas", styles))
    story.append(checklist_table([
        ("Series & DataFrame creation from dict, list, CSV, JSON", "B"),
        ("Selecting data — loc, iloc, boolean filtering", "B"),
        ("Handling missing values — isnull, fillna, dropna", "B"),
        ("Data types — dtypes, astype, pd.to_datetime", "B"),
        ("GroupBy — groupby, agg, transform, apply", "I"),
        ("Merging / joining — merge, join, concat", "I"),
        ("Pivot tables & crosstabs", "I"),
        ("String operations — str accessor methods", "I"),
        ("Datetime operations — dt accessor, resample, rolling", "I"),
        ("Apply & map — row/column-wise transformations", "I"),
        ("MultiIndex DataFrames", "I"),
        ("Memory optimization — categoricals, chunked reading", "A"),
        ("Method chaining & pipe for clean pipelines", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Math Foundations for ML", styles))
    story.append(checklist_table([
        ("Linear Algebra — vectors, matrices, dot products, norms", "B"),
        ("Statistics — mean, median, mode, variance, std dev", "B"),
        ("Probability basics — conditional prob, Bayes theorem", "I"),
        ("Distributions — Normal, Bernoulli, Binomial, Poisson", "I"),
        ("Calculus intuition — gradients, chain rule (for backprop)", "I"),
        ("Eigenvalues & eigenvectors (for PCA)", "I"),
        ("Information theory — entropy, cross-entropy, KL divergence", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 3 — MACHINE LEARNING
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("3.  MACHINE LEARNING (Classical)", styles, "🤖"))
    story.append(SP(2))

    story.append(cat_header("Core Concepts & Workflow", styles))
    story.append(checklist_table([
        ("Supervised vs Unsupervised vs Reinforcement Learning", "B"),
        ("Train / Validation / Test split strategy", "B"),
        ("Overfitting vs Underfitting — bias-variance tradeoff", "B"),
        ("Cross-validation — k-fold, stratified k-fold", "I"),
        ("Feature engineering — encoding, scaling, binning", "I"),
        ("Feature selection — correlation, mutual information, RFE", "I"),
        ("Handling imbalanced datasets — SMOTE, class weights, undersampling", "I"),
        ("Data leakage — what it is and how to avoid it", "I"),
        ("Scikit-learn pipeline (Pipeline, ColumnTransformer)", "I"),
        ("Experiment tracking — MLflow basics", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Algorithms", styles))
    story.append(checklist_table([
        ("Linear Regression — OLS, assumptions, R2, RMSE", "B"),
        ("Logistic Regression — log-odds, sigmoid, decision boundary", "B"),
        ("Decision Trees — CART, Gini, information gain, pruning", "B"),
        ("Random Forest — bagging, feature importance", "I"),
        ("Gradient Boosting — XGBoost, LightGBM (how they differ)", "I"),
        ("Support Vector Machines — kernel trick, C, gamma", "I"),
        ("Naive Bayes — text classification use case", "B"),
        ("K-Nearest Neighbors — distance metrics, curse of dimensionality", "B"),
        ("K-Means Clustering — elbow method, silhouette score", "I"),
        ("PCA — dimensionality reduction, explained variance ratio", "I"),
        ("DBSCAN & hierarchical clustering", "A"),
        ("Anomaly detection — IsolationForest, Z-score methods", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Evaluation Metrics", styles))
    story.append(checklist_table([
        ("Classification — accuracy, precision, recall, F1, ROC-AUC", "B"),
        ("Confusion matrix interpretation", "B"),
        ("Regression — MAE, MSE, RMSE, R2, MAPE", "B"),
        ("Precision-Recall curve vs ROC curve — when to use which", "I"),
        ("Log-loss / cross-entropy loss", "I"),
        ("Multi-class metrics — macro, micro, weighted averages", "I"),
        ("Calibration of probability estimates", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 4 — DEEP LEARNING & NLP
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("4.  DEEP LEARNING & NLP", styles, "🧠"))
    story.append(SP(2))

    story.append(cat_header("Neural Networks Fundamentals", styles))
    story.append(checklist_table([
        ("Perceptron → MLP — layers, neurons, activation functions", "B"),
        ("Forward pass & backpropagation intuition", "B"),
        ("Activation functions — ReLU, Sigmoid, Tanh, Softmax, GELU", "B"),
        ("Loss functions — MSE, cross-entropy, binary cross-entropy", "B"),
        ("Optimizers — SGD, Adam, AdaGrad, RMSProp", "I"),
        ("Batch size, learning rate, epochs — practical tuning", "I"),
        ("Regularization — Dropout, L1/L2, Batch Normalization", "I"),
        ("Weight initialization — Xavier, He", "I"),
        ("PyTorch basics — tensors, autograd, nn.Module, DataLoader", "I"),
        ("Early stopping & learning rate scheduling", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("NLP Foundations", styles))
    story.append(checklist_table([
        ("Text preprocessing — tokenization, lowercasing, stopwords, stemming", "B"),
        ("Bag of Words & TF-IDF representations", "B"),
        ("Word embeddings — Word2Vec, GloVe intuition", "I"),
        ("Recurrent networks — RNN, LSTM, GRU (concept level)", "I"),
        ("Attention mechanism — query, key, value intuition", "I"),
        ("Transformer architecture — encoder, decoder, self-attention", "I"),
        ("Tokenizers — BPE, WordPiece (used in BERT/GPT)", "I"),
        ("Positional encoding", "I"),
        ("Named Entity Recognition (NER), POS tagging basics", "B"),
        ("Text classification pipeline end-to-end", "B"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 5 — GENAI, LLMs & RAG
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("5.  GENERATIVE AI, LLMs & RAG", styles, "✨"))
    story.append(SP(2))

    story.append(cat_header("LLM Fundamentals", styles))
    story.append(checklist_table([
        ("GPT architecture — decoder-only transformer, causal attention", "B"),
        ("BERT architecture — encoder-only, bidirectional attention", "B"),
        ("Pre-training vs fine-tuning vs in-context learning", "B"),
        ("Prompt engineering — zero-shot, few-shot, chain-of-thought", "B"),
        ("System prompt, user prompt, assistant turn structure", "B"),
        ("Temperature, top-p, top-k, max_tokens parameters", "B"),
        ("Hallucination — why it happens, mitigation strategies", "I"),
        ("Instruction tuning & RLHF overview", "I"),
        ("PEFT methods — LoRA, QLoRA (concept + when to use)", "I"),
        ("Context window limitations & chunking strategies", "I"),
        ("Quantization — INT8, INT4, GGUF formats", "I"),
        ("Model evaluation — BLEU, ROUGE, BERTScore, LLM-as-judge", "I"),
        ("Structured outputs / JSON mode from LLMs", "I"),
        ("Function calling / tool use in LLMs", "I"),
        ("Multi-modal LLMs — vision + text (GPT-4V, Gemini Vision)", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("RAG (Retrieval Augmented Generation) — YOUR CORE SKILL", styles))
    story.append(note_box(
        "You already built RAG at TCS! Focus on depth: chunking strategies, re-ranking, evaluation.",
        styles))
    story.append(checklist_table([
        ("RAG pipeline architecture — ingest → embed → store → retrieve → generate", "B"),
        ("Document loaders — PDF, HTML, DOCX, web scraping", "B"),
        ("Text chunking strategies — fixed, recursive, semantic, sentence", "I"),
        ("Chunk overlap — why it matters, optimal settings", "I"),
        ("Embedding models — OpenAI, Vertex AI, sentence-transformers", "B"),
        ("Vector databases — FAISS (in-memory), ChromaDB, Pinecone, Weaviate", "I"),
        ("Similarity search — cosine similarity, dot product, Euclidean", "B"),
        ("ANN indexes — HNSW, IVF (how FAISS scales)", "I"),
        ("Hybrid search — combining dense + sparse (BM25) retrieval", "I"),
        ("Re-ranking — cross-encoder re-rankers (Cohere, sentence-transformers)", "I"),
        ("Metadata filtering in vector DBs", "I"),
        ("Parent-document retrieval & small-to-big chunking", "I"),
        ("HyDE (Hypothetical Document Embeddings)", "A"),
        ("RAG evaluation — faithfulness, answer relevancy, context precision (RAGAS)", "I"),
        ("GraphRAG — entity-based retrieval using knowledge graphs", "A"),
        ("Multi-hop RAG for complex question answering", "A"),
        ("RAG vs fine-tuning — when to use which", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("LLM Frameworks", styles))
    story.append(checklist_table([
        ("LangChain — chains, LCEL syntax, runnable interface", "I"),
        ("LangChain — retrievers, memory, callbacks", "I"),
        ("LangChain — document loaders & text splitters", "I"),
        ("LangSmith — tracing, evaluation, prompt management", "I"),
        ("LlamaIndex — index types, query engine, response synthesizer", "I"),
        ("Haystack — pipeline concept, nodes, evaluation", "I"),
        ("Google Vertex AI SDK — generate_content, embeddings API", "B"),
        ("OpenAI SDK — chat completions, embeddings, function calling", "B"),
        ("Anthropic Claude API — messages format, streaming", "I"),
        ("Semantic Kernel (Microsoft) — basics", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 6 — AGENTIC AI
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("6.  AGENTIC AI & MULTI-AGENT SYSTEMS", styles, "🤖"))
    story.append(SP(2))

    story.append(cat_header("Agent Fundamentals", styles))
    story.append(checklist_table([
        ("Agent loop — observe → think → act → observe (ReAct pattern)", "B"),
        ("Tools / Function calling — registering and handling tool results", "B"),
        ("Planning strategies — Chain-of-Thought, Tree-of-Thoughts, ReAct", "I"),
        ("Memory types — in-context, external (vector DB), episodic, semantic", "I"),
        ("Agent state management across turns", "I"),
        ("Guardrails & safety in agents — output validation, human-in-the-loop", "I"),
        ("Structured output parsing — Pydantic models for tool inputs/outputs", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Google ADK — YOUR CURRENT STACK", styles))
    story.append(note_box(
        "Google ADK is your differentiator. Go deep here — most candidates don't have this.",
        styles))
    story.append(checklist_table([
        ("ADK agent definition — BaseAgent, LlmAgent structure", "B"),
        ("Tools in ADK — FunctionTool, built-in tools", "B"),
        ("Session management & state persistence in ADK", "I"),
        ("Multi-agent patterns in ADK — coordinator + sub-agents", "I"),
        ("Streaming responses in ADK agents", "I"),
        ("Deploying ADK agents to Vertex AI Agent Engine", "I"),
        ("Evaluation framework in ADK", "I"),
        ("ADK + Gemini models — model selection, safety settings", "B"),
        ("ADK callbacks — before/after model, before/after tool", "A"),
        ("ADK custom authentication for tools", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Model Context Protocol (MCP)", styles))
    story.append(checklist_table([
        ("MCP architecture — host, client, server roles", "B"),
        ("MCP primitives — tools, resources, prompts", "B"),
        ("Building an MCP server in Python (FastMCP / official SDK)", "I"),
        ("MCP tool schema — JSON schema, annotations", "I"),
        ("Connecting MCP servers to Claude Desktop / ADK", "I"),
        ("MCP resources — URI templates, list/read resource handlers", "I"),
        ("MCP authentication & transport — SSE vs stdio", "I"),
        ("MCP sampling — allowing server to request LLM completions", "A"),
        ("Production MCP server — error handling, logging, versioning", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Multi-Agent Frameworks", styles))
    story.append(checklist_table([
        ("LangGraph — nodes, edges, state graph, conditional routing", "I"),
        ("LangGraph — human-in-the-loop, persistence, streaming", "I"),
        ("CrewAI — agents, tasks, crew, process (sequential/hierarchical)", "I"),
        ("AutoGen — conversable agents, group chat manager", "I"),
        ("Agent communication patterns — supervisor, peer-to-peer, blackboard", "I"),
        ("Error recovery in multi-agent systems", "A"),
        ("Agent observability — LangSmith, Langfuse tracing", "I"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 7 — FASTAPI & BACKEND
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("7.  FASTAPI & BACKEND ENGINEERING", styles, "⚡"))
    story.append(SP(2))

    story.append(cat_header("FastAPI Core — YOUR CURRENT STACK", styles))
    story.append(checklist_table([
        ("Route decorators — GET, POST, PUT, DELETE", "B"),
        ("Path parameters, query parameters, request body", "B"),
        ("Pydantic models for request/response validation", "B"),
        ("Dependency injection — Depends()", "I"),
        ("Background tasks — BackgroundTasks", "I"),
        ("Async endpoints — async def, await", "I"),
        ("Middleware — CORS, logging, rate limiting", "I"),
        ("Error handling — HTTPException, custom exception handlers", "I"),
        ("File uploads — UploadFile, File", "I"),
        ("Streaming responses — StreamingResponse (for LLM token streaming)", "I"),
        ("WebSockets in FastAPI", "I"),
        ("Authentication — JWT tokens, OAuth2 with FastAPI", "I"),
        ("OpenAPI docs — tags, descriptions, response models", "B"),
        ("Testing — TestClient, pytest fixtures for FastAPI", "I"),
        ("Lifespan events — startup/shutdown (DB connections, model loading)", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("API Design & System Design Basics", styles))
    story.append(checklist_table([
        ("REST principles — statelessness, resource naming conventions", "B"),
        ("HTTP status codes — 200, 201, 400, 401, 403, 404, 422, 500", "B"),
        ("Pagination — offset-based, cursor-based", "I"),
        ("Rate limiting concepts", "I"),
        ("Caching — Redis basics for API response caching", "I"),
        ("Message queues — Pub/Sub, Cloud Tasks basics", "I"),
        ("Microservices vs monolith tradeoffs", "I"),
        ("API versioning strategies", "I"),
        ("System design components — load balancer, DB, cache, queue", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 8 — GCP & MLOps
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("8.  GOOGLE CLOUD PLATFORM & MLOps", styles, "☁️"))
    story.append(SP(2))

    story.append(cat_header("GCP Core Services — YOUR CURRENT STACK", styles))
    story.append(checklist_table([
        ("IAM — roles, service accounts, principle of least privilege", "B"),
        ("Cloud Storage — buckets, object storage, signed URLs, lifecycle", "B"),
        ("BigQuery — dataset/table structure, basic SQL queries", "B"),
        ("BigQuery — partitioning, clustering for performance", "I"),
        ("BigQuery ML (BQML) — training models in SQL", "I"),
        ("Cloud Run — deploying containers, environment variables, scaling", "B"),
        ("Cloud Run — concurrency, min instances, CPU allocation", "I"),
        ("Artifact Registry — storing Docker images", "I"),
        ("Cloud Build — CI/CD pipelines for deployment", "I"),
        ("Secret Manager — storing API keys securely", "I"),
        ("Pub/Sub — topics, subscriptions, push vs pull", "I"),
        ("Cloud Logging & Cloud Monitoring — alerts, dashboards", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Vertex AI — YOUR CURRENT STACK", styles))
    story.append(checklist_table([
        ("Vertex AI Workbench — managed Jupyter notebooks", "B"),
        ("Vertex AI Datasets — creating & managing datasets", "B"),
        ("Vertex AI Training — custom training jobs, pre-built containers", "I"),
        ("Vertex AI Model Registry — versioning, aliases", "I"),
        ("Vertex AI Endpoints — deploying models, traffic splitting", "I"),
        ("Vertex AI Pipelines — Kubeflow Pipelines on Vertex", "I"),
        ("Vertex AI Feature Store", "I"),
        ("Vertex AI Agent Builder — search & conversation apps", "I"),
        ("Vertex AI Agent Engine — deploying ADK agents", "I"),
        ("Gemini API on Vertex — models, safety settings, grounding", "B"),
        ("Embeddings API on Vertex — text-embedding-004 & multimodal", "B"),
        ("Vertex AI Vector Search (Matching Engine)", "I"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Containerization & DevOps", styles))
    story.append(checklist_table([
        ("Docker basics — Dockerfile, build, run, tag, push", "B"),
        ("Docker Compose — multi-service local development", "B"),
        ("Docker best practices — multi-stage builds, .dockerignore", "I"),
        ("Container registries — Artifact Registry, Docker Hub", "B"),
        ("Kubernetes concepts — pods, deployments, services, ingress", "I"),
        ("GKE (Google Kubernetes Engine) basics", "I"),
        ("CI/CD concepts — Git workflow, automated testing, deployment", "I"),
        ("Infrastructure as Code — Terraform basics", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("MLOps Practices", styles))
    story.append(checklist_table([
        ("ML pipeline stages — data validation, training, evaluation, serving", "I"),
        ("Model versioning & lineage tracking", "I"),
        ("Model monitoring — data drift, concept drift detection", "I"),
        ("A/B testing for model evaluation in production", "I"),
        ("MLflow — experiment tracking, model registry", "I"),
        ("Feature stores — purpose, offline vs online serving", "I"),
        ("Continuous training triggers", "A"),
        ("Model explainability — SHAP values, Vertex Explainable AI", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 9 — SQL & DATA ENGINEERING
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("9.  SQL & DATA ENGINEERING", styles, "🗄️"))
    story.append(SP(2))

    story.append(cat_header("SQL Fundamentals to Advanced", styles))
    story.append(checklist_table([
        ("SELECT, WHERE, ORDER BY, LIMIT, DISTINCT", "B"),
        ("Aggregate functions — COUNT, SUM, AVG, MIN, MAX", "B"),
        ("GROUP BY, HAVING", "B"),
        ("JOINs — INNER, LEFT, RIGHT, FULL OUTER, CROSS", "B"),
        ("Subqueries & CTEs (WITH clause)", "I"),
        ("Window functions — ROW_NUMBER, RANK, LAG, LEAD, PARTITION BY", "I"),
        ("String functions — CONCAT, SUBSTR, REPLACE, REGEXP", "I"),
        ("Date/time functions — DATE_DIFF, DATE_TRUNC, EXTRACT", "I"),
        ("CASE WHEN expressions", "I"),
        ("BigQuery-specific — UNNEST arrays, STRUCT, JSON functions", "I"),
        ("Query optimization — EXPLAIN, index usage, partitioning", "A"),
        ("Data modeling — star schema, fact vs dimension tables", "A"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 10 — DSA (Interview Prep)
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("10.  DSA & CODING INTERVIEWS", styles, "💻"))
    story.append(SP(2))
    story.append(note_box(
        "For AI/ML roles at product companies (Google, Microsoft), DSA is tested. For AI-focused startups & TCS-level companies, practical coding matters more.",
        styles))
    story.append(SP(2))

    story.append(cat_header("Data Structures", styles))
    story.append(checklist_table([
        ("Arrays & strings — two pointers, sliding window patterns", "B"),
        ("Hash maps & hash sets — frequency counting, lookups", "B"),
        ("Linked lists — traversal, reversal, fast/slow pointers", "B"),
        ("Stacks & queues — implementation, use cases", "B"),
        ("Binary search — on sorted array and answer space", "I"),
        ("Trees — traversal (BFS/DFS), height, lowest common ancestor", "I"),
        ("Heaps/Priority queues — top-K problems", "I"),
        ("Graphs — BFS, DFS, topological sort, Union-Find", "I"),
        ("Dynamic programming — memoization, tabulation, common patterns", "A"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Complexity & Problem Solving", styles))
    story.append(checklist_table([
        ("Big-O time & space complexity analysis", "B"),
        ("Recognizing patterns — two pointers, sliding window, fast/slow ptr", "I"),
        ("Greedy algorithms", "I"),
        ("Backtracking — permutations, combinations, N-queens", "A"),
        ("Bit manipulation basics", "I"),
        ("LeetCode Easy — solve 30+ problems across key topics", "B"),
        ("LeetCode Medium — solve 20+ problems, especially arrays/trees/graphs", "I"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # SECTION 11 — SOFT SKILLS & PORTFOLIO
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("11.  PROJECTS, PORTFOLIO & COMMUNICATION", styles, "🚀"))
    story.append(SP(2))

    story.append(cat_header("Portfolio Projects (Build These)", styles))
    story.append(checklist_table([
        ("Production-grade RAG app with evaluation (RAGAS metrics)", "I", "Use LangChain/LlamaIndex + FAISS/ChromaDB + FastAPI"),
        ("LangGraph or CrewAI multi-agent system with a real use case", "I", "E.g. research agent, code review agent"),
        ("MCP server for a useful tool (calendar, DB, web search)", "I", "Show ADK + MCP integration"),
        ("ML classification project with full pipeline on Vertex AI", "I", "End-to-end: EDA → training → deployment → monitoring"),
        ("Open-source contribution (even docs/examples) to ADK or LangChain", "A"),
        ("Technical blog post or LinkedIn article on a project you built", "B"),
        ("GitHub profile — pinned repos, README, clean commits", "B"),
    ], styles))
    story.append(SP(2))

    story.append(cat_header("Communication & Interview Skills", styles))
    story.append(checklist_table([
        ("STAR method for behavioral questions (Situation, Task, Action, Result)", "B"),
        ("Ability to explain ML/AI concepts to non-technical stakeholders", "B"),
        ("System design of AI systems — articulate RAG/agent architecture clearly", "I"),
        ("Writing clear technical documentation & ADRs", "I"),
        ("Presenting ML results — metrics, tradeoffs, business impact", "I"),
    ], styles))
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # PAGE BREAK — 2 WEEK PLAN
    # ════════════════════════════════════════════════════════════════════
    story.append(PageBreak())

    story.append(section_header("📅  2-WEEK SKILL AUDIT & LEARNING PLAN", styles))
    story.append(SP(3))
    story.append(note_box(
        "Goal of Week 1: Audit everything you know. Fill checkboxes honestly. Identify real gaps. "
        "Goal of Week 2: Fill the most critical gaps with focused 2-3 hour daily sessions.",
        styles))
    story.append(SP(4))

    week_data = [
        # (Day, Focus Area, Key Actions)
        ("Day 1", "Python Audit", "Work through Python sections 1.1–1.3. Write small scripts to test each concept you're unsure about. No tutorials — just code."),
        ("Day 2", "NumPy & Pandas Audit", "Load a real CSV (your ticket data if possible). Perform groupby, merge, pivot, apply operations. Check boxes as you do each."),
        ("Day 3", "ML Concepts Audit", "Pick a classification dataset (UCI/Kaggle). Build: preprocessing pipeline → model → evaluation. Document gaps."),
        ("Day 4", "RAG Deep Dive", "Re-build your TCS RAG pipeline from scratch locally. Add chunking strategy comparison. Add RAGAS evaluation. Time: 3 hrs."),
        ("Day 5", "ADK + MCP Deep Dive", "Build a fresh ADK agent with a custom MCP server. Document what you understand vs what you had to look up."),
        ("Day 6", "FastAPI + GCP", "Deploy a FastAPI endpoint to Cloud Run from scratch. Add auth, logging, error handling. Time yourself."),
        ("Day 7", "REVIEW & GAP LIST", "Look at all unchecked boxes. Write a prioritized gap list: Top 10 things to learn in Week 2."),
        ("Day 8", "Fill Gap #1-2", "Focus on your 2 highest-priority gaps. Deep study: docs + coding exercise per topic. No videos — hands-on only."),
        ("Day 9", "Fill Gap #3-4", "Same approach. If DSA is a gap: solve 5 LeetCode Easy + 3 Medium problems."),
        ("Day 10", "LLM Internals", "Study transformer architecture (Illustrated Transformer blog). Implement attention from scratch (small version). Solidify."),
        ("Day 11", "MLOps & Vertex AI", "Walk through a Vertex AI Pipeline end-to-end. Add model monitoring. Check those boxes."),
        ("Day 12", "System Design Practice", "Design a RAG system for 1M documents on paper. Then design a multi-agent ticket classifier. Practice explaining aloud."),
        ("Day 13", "Mock Interview Day", "Do 2 mock technical interviews (use Pramp or a friend). 1 coding round + 1 ML concept round."),
        ("Day 14", "Resume & Portfolio Update", "Update resume with confidence. Push 2 clean projects to GitHub. Write 1 LinkedIn post about a project. Apply to 3 target companies."),
    ]

    week_rows = []
    for day, focus, actions in week_data:
        bg = colors.HexColor("#e8eaf6") if "Day 1" <= day <= "Day 7" else colors.HexColor("#fce4ec")
        week_label = "WEEK 1" if day <= "Day 7" else "WEEK 2"
        day_p    = Paragraph(f"<b>{day}</b>", styles['cat_hdr'])
        focus_p  = Paragraph(f"<b>{focus}</b>", styles['role_title'])
        action_p = Paragraph(actions, styles['role_body'])
        week_rows.append([day_p, focus_p, action_p])

    col_w = PAGE_W - 2*MARGIN
    week_tbl = Table(week_rows, colWidths=[col_w*0.12, col_w*0.22, col_w*0.66])
    week_style = [
        ("GRID",          (0,0), (-1,-1), 0.4, C_BORDER),
        ("VALIGN",        (0,0), (-1,-1), "TOP"),
        ("TOPPADDING",    (0,0), (-1,-1), 5),
        ("BOTTOMPADDING", (0,0), (-1,-1), 5),
        ("LEFTPADDING",   (0,0), (-1,-1), 6),
        ("ROWBACKGROUNDS",(0,0), (2,6),   [colors.HexColor("#e8eaf6"), colors.HexColor("#f3f4fb")]),
        ("ROWBACKGROUNDS",(0,7), (2,13),  [colors.HexColor("#fce4ec"), colors.HexColor("#fff0f3")]),
    ]
    week_tbl.setStyle(TableStyle(week_style))
    story.append(week_tbl)
    story.append(SP(5))

    # ════════════════════════════════════════════════════════════════════
    # TARGET ROLES TABLE
    # ════════════════════════════════════════════════════════════════════
    story.append(section_header("🎯  TARGET ROLES FOR COMPANY SWITCH (2025-2026 Market)", styles))
    story.append(SP(3))

    roles = [
        ("AI/ML Engineer\n(GenAI Focused)", "★★★★★\nBest Fit",
         "Your #1 target. 2 years TCS GenAI + RAG + ADK + MCP + Vertex AI is exactly what companies want.",
         "₹18–32 LPA (India) | $130–180K (US/remote)",
         "Deepen LLM internals, RAG evaluation, LangGraph. Build 2 strong portfolio projects."),
        ("MLOps / AI Platform Engineer", "★★★★☆\nStrong Fit",
         "Your deployment skills (Cloud Run, Vertex AI, FastAPI) are relevant. Gap: CI/CD pipelines, Kubeflow.",
         "₹18–28 LPA",
         "Learn Vertex AI Pipelines, Kubeflow, model monitoring deeply."),
        ("Data Scientist", "★★★☆☆\nModerate Fit",
         "You'll need stronger stats, EDA, and classical ML depth. Your AI engineering skills are overkill for pure DS.",
         "₹12–22 LPA",
         "Fill stats gaps, practice EDA storytelling, learn visualization (matplotlib, seaborn, plotly)."),
        ("AI Research Engineer", "★★☆☆☆\nFuture Target",
         "Requires deep ML theory, paper implementation skills. Target after 1-2 more years of depth.",
         "₹25–45 LPA",
         "Read ML papers, implement models from scratch, contribute to open source ML."),
        ("Backend Engineer\n(AI Systems)", "★★★★☆\nSolid Option",
         "Your FastAPI + GCP + Python skills qualify. Position yourself as 'AI-aware backend engineer'.",
         "₹15–25 LPA",
         "Strengthen system design, distributed systems basics, database design."),
    ]

    role_rows = [["Role", "Fit Score", "Why This Fits You", "Salary Range (India)", "What to Build/Learn"]]
    for role, fit, why, sal, learn in roles:
        role_rows.append([
            Paragraph(f"<b>{role}</b>", styles['role_title']),
            Paragraph(fit, styles['note']),
            Paragraph(why, styles['role_body']),
            Paragraph(sal, styles['note']),
            Paragraph(learn, styles['role_body']),
        ])

    col_w = PAGE_W - 2*MARGIN
    role_tbl = Table(role_rows, colWidths=[col_w*0.16, col_w*0.10, col_w*0.26, col_w*0.18, col_w*0.30])
    role_tbl.setStyle(TableStyle([
        ("BACKGROUND",    (0,0), (-1,0),   C_BG_HEADER),
        ("TEXTCOLOR",     (0,0), (-1,0),   C_TEXT_LIGHT),
        ("FONTNAME",      (0,0), (-1,0),   "Helvetica-Bold"),
        ("FONTSIZE",      (0,0), (-1,0),   9),
        ("GRID",          (0,0), (-1,-1),  0.4, C_BORDER),
        ("VALIGN",        (0,0), (-1,-1),  "TOP"),
        ("TOPPADDING",    (0,0), (-1,-1),  5),
        ("BOTTOMPADDING", (0,0), (-1,-1),  5),
        ("LEFTPADDING",   (0,0), (-1,-1),  5),
        ("ROWBACKGROUNDS",(0,1), (-1,-1),  [colors.white, C_ROW_ALT]),
    ]))
    story.append(role_tbl)
    story.append(SP(5))

    # ── CONFIDENCE BUILDING NOTE ────────────────────────────────────────
    story.append(section_header("💡  A NOTE ON CONFIDENCE", styles))
    story.append(SP(3))
    conf_text = (
        "Dinesh — confidence doesn't come from knowing everything. It comes from knowing <i>where you stand</i>. "
        "The impostor syndrome you feel is extremely common among engineers who build real things (vs those who only watch tutorials). "
        "You have deployed RAG pipelines and ADK agents in production at a major company. That is real. "
        "Use this checklist not to feel bad about gaps, but to convert unknown unknowns into known unknowns — "
        "that alone will double your confidence. Check each box only when you can <b>explain it to someone else</b>, "
        "not just when you've read about it. That standard will make you genuinely strong."
    )
    conf_p = Paragraph(conf_text, styles['role_body'])
    conf_tbl = Table([[conf_p]], colWidths=[PAGE_W - 2*MARGIN])
    conf_tbl.setStyle(TableStyle([
        ("BACKGROUND",    (0,0), (-1,-1), C_NOTE_BG),
        ("BOX",           (0,0), (-1,-1), 1, C_NOTE_BORDER),
        ("TOPPADDING",    (0,0), (-1,-1), 10),
        ("BOTTOMPADDING", (0,0), (-1,-1), 10),
        ("LEFTPADDING",   (0,0), (-1,-1), 12),
        ("RIGHTPADDING",  (0,0), (-1,-1), 12),
    ]))
    story.append(conf_tbl)

    # ── FOOTER ──────────────────────────────────────────────────────────
    story.append(SP(6))
    story.append(HRFlowable(width="100%", thickness=0.5, color=C_BORDER))
    story.append(SP(2))
    story.append(Paragraph(
        "Dinesh Sheelam — AI/ML Skill Mastery Checklist  |  Generated June 2026  |  Beginner → Expert Coverage",
        styles['footer']))

    doc.build(story)
    print(f"PDF saved to {path}")

build_pdf("Dinesh_AI_ML_Skill_Checklist.pdf")
```
