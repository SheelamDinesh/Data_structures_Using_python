```python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []

def addChild(parent, child):
    parent.children.append(child)


def print_all_data(node):
    for child in node.children:
        print(str(node.data) + " -> " + str(child.data))
        print_all_data(child)

def print_all_leaf_data(node):
    childnodes = ', '.join(str(child.data) for child in node.children)
    print(str(node.data) + " -> " + str(childnodes))
    for child in node.children:
        print_all_leaf_data(child)

def print_only_leaf_data(node):
    if len(node.children) == 0:
        print(node.data)
    for child in node.children:
        #print(str(node.data) + " -> " + str(child.data))
        print_only_leaf_data(child)

def print_degree_ofNode(node, parent):
    if parent == None:
        print(str(node.data) + " -> " + str(len(node.children)))
    else:
        print(str(node.data) + " -> " + str(len(node.children)+1))
    for child in node.children:
        print_degree_ofNode(child, node)
```

```python
root_node = Node(1)
n2 = Node(2)
n3 = Node(3)
n4 = Node(4)
n5 = Node(5)
```

```python
addChild(root_node, n2)
addChild(root_node, n3)
addChild(n2, n4)
addChild(n2, n5)
```
